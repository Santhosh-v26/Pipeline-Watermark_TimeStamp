```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

### Top 10 Customers by Total Spend

```python
spark.table('main.gold_task.customers_order_summary')\
    .orderBy(col('total_spent').desc())\
        .limit(10).show()
```

### Top 3 Products by Revenue

```python
from pyspark.sql.window import Window
w = Window.partitionBy("category_name").orderBy(col("total_revenue").desc())

spark.table("main.gold_task.product_sale_summary") \
     .withColumn("rn", row_number().over(w)) \
     .filter(col("rn") <= 3) \
     .orderBy("category_name", col("total_revenue").desc()) \
     .show()
```

### Customer Payment Summary

```python
gold3_df = spark.table("main.gold_task.customer_payment_summary").orderBy(col('total_paid').desc())
gold3_df.show()
```
