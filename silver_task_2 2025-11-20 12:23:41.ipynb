```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
silver_df = ['customers','orders','products','payments','categories']
for table in silver_df:
    df = spark.read.table(f'main.raw_task.{table}')


    numeric_col = [c for c in df.columns if 'amount' in c.lower() or 'price' in c.lower()]
    '''
    numeric_col[]
    for c in df.columns:
        if df.columns  == 'amount'.lower() or df.colums = 'price'.lower():
        numeric_col.append(c)
    '''
    for column in numeric_col:
        df = df.withColumn(column, col(column).cast('double'))

    pk = table[:-1]+'_id'
    if pk in df.columns:
        df = df.filter(col(pk).isNotNull())

    df.write.mode('overwrite').saveAsTable(f'main.silver_task.{table}')
```
